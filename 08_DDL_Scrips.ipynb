```sql
--Create database schemas

CREATE SCHEMA IF NOT EXISTS incubyte_databricks.raw;

CREATE SCHEMA IF NOT EXISTS incubyte_databricks.staging;

CREATE SCHEMA IF NOT EXISTS incubyte_databricks.gold;

CREATE SCHEMA IF NOT EXISTS incubyte_databricks.audit;
```

```sql
--Create raw member landing table
CREATE TABLE IF NOT EXISTS incubyte_databricks.raw.member_landing
(
    Record_Type STRING,
    Member_Name STRING,
    Member_ID STRING,
    Enrollment_Date STRING,
    Last_Flight_Date STRING,
    Tier_Code STRING,
    Agent_Name STRING,
    State STRING,
    Country STRING,
    Post_Code STRING,
    DOB STRING,
    Is_Active STRING,
    Source_File STRING,
    Source_Country STRING,
    Ingested_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create staging member table
CREATE TABLE IF NOT EXISTS incubyte_databricks.staging.member_staging
(
    Member_Name STRING,
    Member_ID STRING,
    Enrollment_Date DATE,
    Last_Flight_Date DATE,
    Tier_Code STRING,
    Agent_Name STRING,
    State STRING,
    Country STRING,
    Post_Code STRING,
    DOB DATE,
    Is_Active STRING,
    Age INT,
    Stale_Member BOOLEAN,
    Enrollment_Date_Status STRING,
    Last_Flight_Date_Status STRING,
    DOB_Status STRING,
    Source_Country STRING,
    Ingested_At TIMESTAMP,
    Processed_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create Australia Target Table
CREATE TABLE IF NOT EXISTS incubyte_databricks.gold.Table_Australia
(
    Member_Name STRING,
    Member_ID STRING,
    Enrollment_Date DATE,
    Last_Flight_Date DATE,
    Tier_Code STRING,
    Agent_Name STRING,
    State STRING,
    Country STRING,
    Post_Code STRING,
    DOB DATE,
    Is_Active STRING,
    Age INT,
    Stale_Member BOOLEAN,
    Source_Country STRING,
    Ingested_At TIMESTAMP,
    Updated_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create India Target Table
CREATE TABLE IF NOT EXISTS incubyte_databricks.gold.Table_India
(
    Member_Name STRING,
    Member_ID STRING,
    Enrollment_Date DATE,
    Last_Flight_Date DATE,
    Tier_Code STRING,
    Agent_Name STRING,
    State STRING,
    Country STRING,
    Post_Code STRING,
    DOB DATE,
    Is_Active STRING,
    Age INT,
    Stale_Member BOOLEAN,
    Source_Country STRING,
    Ingested_At TIMESTAMP,
    Updated_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create USA Target Table
CREATE TABLE IF NOT EXISTS incubyte_databricks.gold.Table_USA
(
    Member_Name STRING,
    Member_ID STRING,
    Enrollment_Date DATE,
    Last_Flight_Date DATE,
    Tier_Code STRING,
    Agent_Name STRING,
    State STRING,
    Country STRING,
    Post_Code STRING,
    DOB DATE,
    Is_Active STRING,
    Age INT,
    Stale_Member BOOLEAN,
    Source_Country STRING,
    Ingested_At TIMESTAMP,
    Updated_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create Redemtion transaction table\

CREATE TABLE IF NOT EXISTS incubyte_databricks.gold.redemption_transactions
(
    Member_ID STRING,
    Feed_Date DATE,
    Transaction_ID STRING,
    Transaction_Date DATE,
    Partner STRING,
    Miles_Redeemed BIGINT,
    Transaction_Status STRING,
    Source_File STRING,
    Ingested_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create rejecrt record table

CREATE TABLE IF NOT EXISTS incubyte_databricks.audit.rejected_member_records
(
    Member_Name STRING,
    Member_ID STRING,
    Source_Country STRING,
    Enrollment_Date DATE,
    Rejection_Reason STRING,
    Rejected_At TIMESTAMP
)
USING DELTA;
```

```sql
--Create data quality summary table
CREATE TABLE IF NOT EXISTS incubyte_databricks.audit.data_quality_summary
(
    Dataset_Name STRING,
    Total_Records BIGINT,
    Valid_Records BIGINT,
    Invalid_Records BIGINT,
    Missing_Records BIGINT,
    Validation_Date TIMESTAMP
)
USING DELTA;
```
